```python
#Q1
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

corrected_temperature = temperature + 2
print("Corrected Temperature:", corrected_temperature)

fahrenheit = corrected_temperature * 9/5 + 32
print("Fahrenheit:", fahrenheit)

greater_than_32 = corrected_temperature[corrected_temperature > 32]
print("Readings greater than 32°C:", greater_than_32)

count = np.sum(corrected_temperature > 32)
print("Count:", count)

print("Vectorization performs operations on the entire array at once instead of processing each element separately.")
print("Boolean indexing directly selects values satisfying a condition, making the code faster and simpler than loops.")
```

```python
#Q2
import numpy as np

steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

total_steps = np.sum(steps)
print("Total Steps:", total_steps)

mean_steps = np.mean(steps)
print("Mean Steps:", mean_steps)

maximum = np.max(steps)
minimum = np.min(steps)

print("Maximum:", maximum)
print("Minimum:", minimum)

day_total = np.sum(steps, axis=0)
print("Total Steps Per Day:", day_total)

user_total = np.sum(steps, axis=1)
print("Total Steps Per User:", user_total)

position = np.unravel_index(np.argmax(steps), steps.shape)

print("Position of Maximum:", position)
print("Maximum Steps:", steps[position])
```

```python
#Q3
import numpy as np

original = np.array([1, 2, 3, 4, 5, 6])
print("Original:", original)

subset = original[1:5]
print("Subset:", subset)

subset[0] = 999

print("Original after modifying subset:", original)
print("Subset:", subset)

copied = original[1:5].copy()
copied[0] = 500

print("Original after modifying copy:", original)
print("Copied Array:", copied)

matrix = np.arange(1, 13).reshape(3, 4)
print("Matrix:\n", matrix)

print("First Row:", matrix[0])
print("Last Row:", matrix[-1])
print("Second Column:", matrix[:, 1])
print("Rows 1-2 and Columns 2-3:\n", matrix[1:3, 2:4])

flat_array = matrix.flatten()
ravel_array = matrix.ravel()

print("Flatten:", flat_array)
print("Ravel:", ravel_array)

ravel_array[0] = 100

print("Ravel after modification:", ravel_array)
print("Matrix after modifying ravel:\n", matrix)

flat_array[1] = 200

print("Flatten after modification:", flat_array)
print("Matrix after modifying flatten:\n", matrix)

print("Shape:", matrix.shape)
print("Dimensions:", matrix.ndim)
print("Size:", matrix.size)
print("Data Type:", matrix.dtype)
```

```python
#Q4
import numpy as np

X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

print("Shape of X:", X.shape)
print("Dimensions of X:", X.ndim)

XT = X.T
print("Transpose:\n", XT)

print("Transpose converts the features from columns into rows.")

XT_X = X.T @ X
print("X.T @ X:\n", XT_X)

XT_X_inv = np.linalg.inv(XT_X)
print("Inverse:\n", XT_X_inv)

beta = np.linalg.inv(X.T @ X) @ X.T @ y
print("Coefficients:", beta)

print("The coefficients represent the contribution of sleep hours, activity level and stress level to the assistance score.")

new_user = np.array([5, 40, 7])

prediction = new_user @ beta
print("Predicted Assistance Score:", prediction)
```
