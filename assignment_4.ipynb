```python
#Q1
import pandas as pd

roll_number = input("Enter your college roll number: ")

fixed_entries = [
    {
        "question": "what is the annual fee",
        "answer": "The annual fee is Rs 500.",
        "keywords": "fee cost price charge",
        "category": "billing"
    },
    {
        "question": "how to reset password",
        "answer": "Go to Settings > Reset Password.",
        "keywords": "password reset login",
        "category": "account"
    },
    {
        "question": "what are your working hours",
        "answer": "We are open 9 AM to 5 PM.",
        "keywords": "hours timing open time",
        "category": "general"
    },
    {
        "question": "how can i pay the fee",
        "answer": "You can pay via UPI, card, or net banking.",
        "keywords": "pay payment upi fee",
        "category": "billing"
    }
]

categories = ["billing", "account", "general"]

digits = [int(ch) for ch in roll_number if ch.isdigit()]

last_two = digits[-2:]

personal_entries = []

for d in last_two:
    category = categories[d % 3]

    if category == "billing":
        entry = {
            "question": "how can i check my pending payment",
            "answer": "You can check pending payments from the billing section of your account.",
            "keywords": "pending payment bill",
            "category": "billing"
        }

    elif category == "account":
        entry = {
            "question": "how do i update my registered mobile number",
            "answer": "Go to account settings and update your registered mobile number.",
            "keywords": "mobile update account",
            "category": "account"
        }

    else:
        entry = {
            "question": "where can i find general information",
            "answer": "General information is available in the help and support section.",
            "keywords": "help information support",
            "category": "general"
        }

    personal_entries.append(entry)

all_entries = fixed_entries + personal_entries

df = pd.DataFrame(all_entries)

print(df)
```

```python
#Q2
def score_query(query, df):
    query_words = set(query.lower().split())

    results = []

    for index, row in df.iterrows():

        faq_text = (
            row["question"].lower() + " " +
            row["keywords"].lower()
        )

        faq_words = set(faq_text.split())

        score = len(query_words.intersection(faq_words))

        if score > 0:
            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "score": score
            })

    result_df = pd.DataFrame(results)

    if len(result_df) > 0:
        result_df = result_df.sort_values(
            by="score",
            ascending=False
        )

    return result_df
```

```python
query = input("Enter your question: ")

results = score_query(query, df)

print(results)
```

```python
#Q3
def same_category(category_name, df):
    return df[df["category"] == category_name]["question"]

personal_category = personal_entries[0]["category"]

print("Category:", personal_category)
print()

print(same_category(personal_category, df))
```

```python
#Q6
def best_match(query, df):

    query_words = set(query.lower().split())

    results = []

    for index, row in df.iterrows():

        faq_text = (
            row["question"].lower() + " " +
            row["keywords"].lower()
        )

        faq_words = set(faq_text.split())

        score = len(query_words.intersection(faq_words))

        if score > 0:
            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "score": score
            })

    if len(results) == 0:
        print("No matching FAQ found.")
        return

    result_df = pd.DataFrame(results)

    highest_score = result_df["score"].max()

    best_results = result_df[
        result_df["score"] == highest_score
    ]

    if len(best_results) > 1:
        print("Multiple equally good matches found:")
    else:
        print("Best match found:")

    print(best_results)
```

```python
best_match("fee", df)
```

```python
best_match("reset password", df)
```
